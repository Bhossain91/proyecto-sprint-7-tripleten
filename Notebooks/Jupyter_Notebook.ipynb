```python
import pandas as pd
import plotly.graph_objects as go
```

```python
df = pd.read_csv('vehicles_us.csv')

df.head()
```

```python
#Info de dataframe
df.info()
```

```python
#Estadisticas
df.describe()
```

```python
#Vaidar valores nulos
df.isnull().sum().sort_values(ascending=False)
```

```python
#Validar duplicados
df.duplicated().sum()
```

```python
#Extraer el fabricante
df['manufacturer'] = df['model'].str.split().str[0]
```

```python
#Frecuencia por fabricante
df['manufacturer'].value_counts()
```

```python
#tipos de vehiculos
df['type'].value_counts()
```

```python
#Condición 
df['condition'].value_counts()
```

```python
#Precio promedio por fabricante
df.groupby('manufacturer')['price'].mean().sort_values(ascending=False)
```

```python
#odemeter promedio por fabricante
df.groupby('manufacturer')['odometer'].mean()
```

```python
#Grafico de cantidad de vehiculos por fabricante
manufacturer_count = (
    df.groupby("manufacturer")
      .size()
      .reset_index(name="count")
)

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=manufacturer_count["manufacturer"],
        y=manufacturer_count["count"],
        text=manufacturer_count["count"],
        textposition="outside",
        name="Vehículos"
    )
)

fig.update_layout(
    title="Cantidad de vehículos por fabricante",
    xaxis_title="Fabricante",
    yaxis_title="Cantidad",
    template="plotly_white"
)

fig.show()
```

```python
#Cantidad de vehiculo por tipo 
vehicle_type = (
    df.groupby(["manufacturer", "type"])
      .size()
      .reset_index(name="count")
)

fig = go.Figure()

for vehicle in vehicle_type["type"].unique():

    temp = vehicle_type[vehicle_type["type"] == vehicle]

    fig.add_trace(
        go.Bar(
            x=temp["manufacturer"],
            y=temp["count"],
            name=vehicle
        )
    )

fig.update_layout(
    title="Cantidad de vehículos por fabricante y tipo",
    xaxis_title="Fabricante",
    yaxis_title="Cantidad",
    barmode="group",
    template="plotly_white"
)

fig.show()
```

```python
#Histograma de Condición de vehiculo por modelo
fig = go.Figure()

for condition in df["condition"].unique():

    temp = df[df["condition"] == condition]

    fig.add_trace(
        go.Histogram(
            x=temp["model_year"],
            name=condition,
            opacity=0.75
        )
    )

fig.update_layout(
    title="Distribución del año del modelo según la condición",
    xaxis_title="Año modelo",
    yaxis_title="Cantidad",
    barmode="overlay",
    template="plotly_white"
)

fig.show()
```

```python
#Grafico de distribución de precios vs odometro
fig = go.Figure()

for manufacturer in df["manufacturer"].unique():

    temp = df[df["manufacturer"] == manufacturer]

    fig.add_trace(
        go.Scatter(
            x=temp["odometer"],
            y=temp["price"],
            mode="markers",
            name=manufacturer
        )
    )

fig.update_layout(
    title="Precio vs Odómetro por fabricante",
    xaxis_title="Odómetro",
    yaxis_title="Precio (USD)",
    template="plotly_white"
)

fig.show()
```

```python
#Comaparación de distribucion de precios entre fabricantes
manufacturer1 = "Ford"
manufacturer2 = "Toyota"
df_compare = df[
    df["manufacturer"].isin([manufacturer1, manufacturer2])
]
fig = go.Figure()

for manufacturer in [manufacturer1, manufacturer2]:

    temp = df_compare[df_compare["manufacturer"] == manufacturer]

    fig.add_trace(
        go.Histogram(
            x=temp["price"],
            histnorm="percent",
            name=manufacturer,
            opacity=0.75
        )
    )

fig.update_layout(
    title="Distribución porcentual de precios",
    xaxis_title="Precio (USD)",
    yaxis_title="Porcentaje",
    barmode="overlay",
    template="plotly_white"
)

fig.show()
```
